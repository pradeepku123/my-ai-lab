## 1. Traditional vs. Semantic Metrics
Let's compare how traditional n-gram overlap (ROUGE) compares to semantic similarity (BERTScore).

```python
reference = ["The cat sat on the mat."]
candidate_1 = ["The cat is sitting on the mat."] # Similar
candidate_2 = ["A feline was resting on the rug."] # Same meaning, different words

# ── ROUGE Score ──
scorer = rouge_scorer.RougeScorer(['rouge1', 'rougeL'], use_stemmer=True)
r1 = scorer.score(reference[0], candidate_1[0])
r2 = scorer.score(reference[0], candidate_2[0])

print(f"ROUGE-L (Candidate 1): {r1['rougeL'].fmeasure:.4f}")
print(f"ROUGE-L (Candidate 2): {r2['rougeL'].fmeasure:.4f}  <-- Very low despite same meaning")

# ── BERTScore ──
P1, R1, F1_1 = bert_score(candidate_1, reference, lang="en", verbose=False)
P2, R2, F1_2 = bert_score(candidate_2, reference, lang="en", verbose=False)

print(f"BERTScore (Candidate 1): {F1_1.item():.4f}")
print(f"BERTScore (Candidate 2): {F1_2.item():.4f}  <-- Much higher! Captures semantics")
```

## 2. RAG Evaluation (The RAGAS Triad)
In Retrieval-Augmented Generation, we evaluate three things:
1. **Faithfulness:** Is the answer grounded in the context? (No hallucinations)
2. **Answer Relevancy:** Does it answer the user's question?
3. **Context Precision:** Was the retrieved context actually useful?
